## What is a Function in Python?

A function is a block of organized, reusable code that is used to perform a single, related action. Functions provide better modularity for your application and a high degree of code reusing.

**Key benefits of using functions:**
*   **Reusability:** Write code once and use it multiple times.
*   **Modularity:** Break down complex problems into smaller, manageable pieces.
*   **Readability:** Make your code easier to understand and maintain.
*   **Reduced Redundancy:** Avoid repeating the same code block.


### Defining a Function

In Python, you define a function using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`.

```python
def function_name(parameters):
    """Docstring (optional): Explains what the function does."""
    # function body - code to be executed
    return expression # (optional) returns a value
```

*   **`def`**: Keyword to declare a function.
*   **`function_name`**: A unique name for the function (follow Python's naming conventions).
*   **`parameters`**: (Optional) Values passed into the function.
*   **`:`**: Marks the end of the function header.
*   **`"""Docstring"""`**: (Optional) A string literal used to document a function's purpose. It's good practice to include one.
*   **Function Body**: Indented block of statements that the function executes.
*   **`return`**: (Optional) Keyword to send a result back to the caller. If omitted, the function implicitly returns `None`.


### Example: A Simple Greeting Function

Let's create a function that takes a name as input and returns a greeting message.

````python
# Define a function called 'greet'
def greet(name):
    """This function greets the person passed in as a parameter."""
    message = f"Hello, {name}! How are you today?"
    return message

# Call the function and store the result
my_greeting = greet('Alice')

# Display the result
display(my_greeting)

# Call the function again with a different name
display(greet('Bob'))
````

#### Explanation of the Example:

1.  **`def greet(name):`**: We define a function named `greet` that accepts one parameter, `name`.
2.  **`"""Docstring..."""`**: This is a docstring explaining what the function does.
3.  **`message = f"Hello, {name}! How are you today?"`**: Inside the function, we create a personalized greeting message using an f-string.
4.  **`return message`**: The function returns the `message` string.
5.  **`my_greeting = greet('Alice')`**: We *call* or *invoke* the `greet` function, passing the string `'Alice'` as an argument for the `name` parameter. The returned value is then assigned to the variable `my_greeting`.
6.  **`display(my_greeting)`**: We display the content of `my_greeting`.
7.  **`display(greet('Bob'))`**: We call the function again, demonstrating its reusability, and directly display the returned value.

````python
# Python Functions: Different Types of Arguments

# --- 1. Positional Arguments ---
# Arguments passed to a function in the correct positional order.
# The order of arguments matters.

def describe_person(name, age, city):
    print(f"Name: {name}, Age: {age}, City: {city}")

print("\n--- Positional Arguments Example ---")
describe_person("Alice", 30, "New York") # Arguments are matched by position
describe_person("Bob", 25, "London")

# --- 2. Keyword Arguments ---
# Arguments passed to a function using their parameter names.
# The order of keyword arguments does not matter.

def print_details(item, quantity, price):
    print(f"Item: {item}, Quantity: {quantity}, Price: ${price:.2f}")

print("\n--- Keyword Arguments Example ---")
# Arguments passed by keyword, order doesn't matter
print_details(item="Laptop", price=1200.50, quantity=1)
print_details(quantity=5, item="Pen", price=1.99)

# --- 3. Default Arguments ---
# Parameters that have a default value if no argument is provided during the function call.
# Default arguments must be placed after any non-default (positional) arguments.

def greet_user(name, message="Hello"):
    print(f"{message}, {name}!")

print("\n--- Default Arguments Example ---")
# Calling the function without providing the 'message' argument (uses default)
greet_user("Charlie")

# Calling the function and overriding the default 'message'
greet_user("David", "Good morning")

def calculate_discount(price, discount_percentage=10):
    discount_amount = price * (discount_percentage / 100)
    final_price = price - discount_amount
    print(f"Original Price: ${price:.2f}, Discount: {discount_percentage}%, Final Price: ${final_price:.2f}")

# Using the default discount percentage (10%)
calculate_discount(100)

# Providing a custom discount percentage (20%)
calculate_discount(200, 20)

print("\n--- Combined Positional, Keyword, and Default Arguments ---")
# You can combine them, but positional arguments must come first,
# followed by keyword arguments (which can include those with defaults).

def configure_settings(user, theme="dark", notifications=True):
    print(f"User: {user}, Theme: {theme}, Notifications: {notifications}")

configure_settings("User1") # Positional for 'user', defaults for 'theme' and 'notifications'
configure_settings("User2", theme="light") # Positional for 'user', keyword for 'theme', default for 'notifications'
configure_settings("User3", notifications=False, theme="blue") # Positional for 'user', keywords for 'notifications' and 'theme'
````

## Lambda Functions (Anonymous Functions)

### Theory:

A `lambda` function is a small, anonymous function in Python. It can take any number of arguments, but can only have one expression. Unlike a normal function defined with the `def` keyword, a `lambda` function doesn't have a name, which is why it's called "anonymous".

**Key Characteristics:**
*   **Anonymous:** They don't have a name.
*   **Single Expression:** They can only contain one expression, which is implicitly returned.
*   **Concise:** Often used for short, throwaway functions that are not going to be reused.
*   **Commonly used:** With higher-order functions like `map()`, `filter()`, `sorted()`, and `reduce()`.

### Syntax:

```python
lambda arguments: expression
```

*   `lambda`: The keyword used to define an anonymous function.
*   `arguments`: Any number of arguments, separated by commas (e.g., `x`, `x, y`, `a, b, c`).
*   `: (colon)`: Separates the arguments from the expression.
*   `expression`: A single expression that is evaluated and returned.

### Examples:

#### 1. Simple Lambda Function

Let's create a lambda function that adds two numbers.

```python
# Using a regular function
def add_regular(x, y):
    return x + y

print(f"Regular function add(2, 3): {add_regular(2, 3)}")

# Using a lambda function
add_lambda = lambda x, y: x + y

print(f"Lambda function add(2, 3): {add_lambda(2, 3)}")

# A lambda to square a number
square = lambda x: x * x
print(f"Lambda function square(5): {square(5)}")
```

#### 2. Lambda with `map()`

The `map()` function applies a given function to each item of an iterable (like a list) and returns an iterator of the results.

```python

# Use map() with a lambda to square each number
squared_numbers = list(map(lambda x: x * x, numbers))
print(f"Original numbers: {numbers}")
print(f"Squared numbers (using map + lambda): {squared_numbers}")

# Use map() with a lambda to double each number
doubled_numbers = list(map(lambda x: x * 2, numbers))
print(f"Doubled numbers (using map + lambda): {doubled_numbers}")
```

#### 3. Lambda with `filter()`

The `filter()` function constructs an iterator from elements of an iterable for which a function returns true.

```python

# Use filter() with a lambda to get even numbers

print(f"Original numbers: {numbers}")

# Use filter() with a lambda to get numbers greater than 5

print(f"Numbers greater than 5 (using filter + lambda): {greater_than_5}")
```

#### 4. Lambda with `sort()` or `sorted()`

Lambda functions are often used as the `key` argument in `sort()` or `sorted()` to specify a custom sorting order.

```python

    {'name': 'Alice', 'age': 20, 'grade': 'B'},

print(f"Sorted by grade: {sorted_by_grade}")
```


## The `map()` Function

### Theory:

The `map()` function is a built-in Python function that applies a given function to each item of an iterable (like a list or tuple) and returns a **map object** (which is an iterator). This map object can then be converted into another iterable type, such as a list or tuple, to see its contents.

`map()` is very useful for performing the same operation on all elements of a sequence without explicitly writing a `for` loop, leading to more concise and often more performant code.

**Key Characteristics:**
*   **Transformation:** It's primarily used to transform data by applying a function to each item.
*   **Lazy Evaluation:** It returns an iterator, meaning it processes elements one by one as they are requested, rather than creating a new list in memory all at once. This is memory-efficient for large datasets.
*   **Function as Argument:** It takes another function as its first argument, making it a higher-order function.
*   **Multiple Iterables:** It can take one or more iterables.

### Syntax:

```python
map(function, iterable, ...)
```

*   `function`: The function to which `map()` passes each item of the iterable(s). It can be a built-in function, a user-defined function, or a `lambda` function.
*   `iterable`: One or more iterable(s) (e.g., list, tuple, string) whose items are passed to the `function`.
*   `...`: Optional additional iterables. If multiple iterables are provided, the `function` must accept that many arguments.

### Examples:

#### 1. Basic `map()` with a User-Defined Function

Let's square each number in a list using a regular function.

```python
def square_number(x):
    return x * x

numbers = [1, 2, 3, 4, 5]

# Apply the 'square_number' function to each item in 'numbers'
squared_numbers_map = map(square_number, numbers)

# The result is a map object, convert it to a list to see values
squared_numbers_list = list(squared_numbers_map)

print(f"Original numbers: {numbers}")
print(f"Squared numbers (using map with def function): {squared_numbers_list}")
```

#### 2. `map()` with a `lambda` Function (Common Use Case)

This is a very common pattern where `lambda` functions are used for quick, inline transformations with `map()`.

```python
numbers = [10, 20, 30, 40, 50]

# Use map() with a lambda to double each number
doubled_numbers_list = list(map(lambda x: x * 2, numbers))

print(f"Original numbers: {numbers}")
print(f"Doubled numbers (using map + lambda): {doubled_numbers_list}")

# Convert temperatures from Celsius to Fahrenheit (F = C * 9/5 + 32)
celsius_temps = [0, 10, 20, 30]
fahrenheit_temps = list(map(lambda c: (c * 9/5) + 32, celsius_temps))

print(f"Celsius temperatures: {celsius_temps}")
print(f"Fahrenheit temperatures: {fahrenheit_temps}")
```

#### 3. `map()` with Multiple Iterables

When `map()` receives multiple iterables, the function provided must accept a corresponding number of arguments. `map()` stops when the shortest iterable is exhausted.

```python
list1 = [1, 2, 3]
list2 = [10, 20, 30]

# Add corresponding elements from two lists
sum_lists = list(map(lambda x, y: x + y, list1, list2))
print(f"List 1: {list1}")
print(f"List 2: {list2}")
print(f"Sum of corresponding elements: {sum_lists}")

# Example with different length iterables (map stops at shortest)
list_a = [1, 2, 3, 4, 5]
list_b = [10, 20, 30]
multiplied_lists = list(map(lambda a, b: a * b, list_a, list_b))
print(f"List A: {list_a}")
print(f"List B: {list_b}")
print(f"Product of corresponding elements (stops at shortest): {multiplied_lists}")
```

#### 4. `map()` with Built-in Functions

You can also pass built-in functions to `map()`.

```python
words = ['apple', 'banana', 'cherry']

# Get the length of each word
word_lengths = list(map(len, words))
print(f"Words: {words}")
print(f"Lengths of words: {word_lengths}")

# Convert strings to integers

integer_numbers = list(map(int, string_numbers))
print(f"String numbers: {string_numbers}")
print(f"Integer numbers: {integer_numbers}")
```

*   **Selection:** It's primarily used to select items from an iterable that meet a certain condition.

*   **Equivalent to a comprehension:** `filter(function, iterable)` is equivalent to `(item for item in iterable if function(item))` or `[item for item in iterable if function(item)]`.

*   `iterable`: The iterable (e.g., list, tuple, string) whose elements are to be filtered.

## Python Modules: An In-Depth Explanation

### Theory:

A **module** in Python is simply a file containing Python definitions and statements. The file name is the module name with the suffix `.py` appended. Modules are a fundamental concept for organizing Python code into reusable units.

Think of a module as a toolbox. Each tool (function, class, variable) in the toolbox is designed for a specific task. When you need a particular tool, you don't need to carry the whole toolbox; you just import the specific tool or the entire toolbox into your current workspace.

**Why use Modules? (Benefits):**

1.  **Code Organization:** Modules help in structuring large programs logically, grouping related code into separate files. This makes the code easier to manage and understand.
2.  **Reusability:** Code written in one module can be reused in multiple other programs or modules without rewriting it. This promotes the DRY (Don't Repeat Yourself) principle.
3.  **Namespace Management:** Modules define their own namespaces. This prevents naming conflicts between identifiers (variables, functions, classes) in different modules. For example, two different modules can have a function named `add()` without conflicting.
4.  **Maintainability:** Changes made in one module are localized and less likely to affect other parts of a large application, making maintenance easier.
5.  **Distribution:** Modules are the building blocks for creating Python packages, which can then be easily shared and installed by others.

### Creating a Module:

To create a module, you simply write your Python code and save it in a file with a `.py` extension. For example, if you create a file named `my_utilities.py` with some functions, `my_utilities` becomes a module.

Let's imagine we have a file named `calculations.py` with the following content:

```python
# calculations.py

def add(a, b):
    """Returns the sum of two numbers."""
    return a + b

def subtract(a, b):
    """Returns the difference of two numbers."""
    return a - b

def multiply(a, b):
    """Returns the product of two numbers."""
    return a * b

def divide(a, b):
    """Returns the division of two numbers, handles division by zero."""
    if b == 0:
        return "Error: Cannot divide by zero"
    return a / b

PI = 3.14159
EULER_NUMBER = 2.71828

print("Calculations module loaded!") # This will run when the module is imported
```

### Importing a Module:

There are several ways to import and use modules in Python:

#### 1. `import module_name`

This is the most common way to import a module. It imports the entire module, and you access its contents using the module name followed by a dot (`.`).

**Syntax:**

```python
import module_name
```

**Example:** (Assuming `calculations.py` exists in the same directory or Python's path)

```python
# First, let's create the 'calculations.py' file in a temporary location
# In a real scenario, this file would exist on your system.
import os

module_code = """
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: Cannot divide by zero"
    return a / b

PI = 3.14159
EULER_NUMBER = 2.71828

print("Calculations module loaded!")
"""

with open('calculations.py', 'w') as f:
    f.write(module_code)

# Now, import the module
import calculations

print(f"Sum: {calculations.add(10, 5)}")
print(f"Difference: {calculations.subtract(10, 5)}")
print(f"Product: {calculations.multiply(10, 5)}")
print(f"Division: {calculations.divide(10, 2)}")
print(f"Value of PI: {calculations.PI}")

# Clean up the created file (optional)
os.remove('calculations.py')
```

#### 2. `import module_name as alias`

This method allows you to give the imported module an alias (a shorter or different name) for convenience. This is especially useful for long module names or to avoid naming conflicts if you import multiple modules with similar names.

**Syntax:**

```python

```

**Example:**


### Where Python Looks for Modules:

When you import a module, Python searches for it in the following order:

1.  The directory containing the input script (or the current working directory).
2.  The list of directories specified in the `PYTHONPATH` environment variable.
3.  The installation-dependent default directories (e.g., `site-packages`).

You can inspect the search path using `sys.path`:

```python
import sys
print(sys.path)
```

## Python Standard Library: Commonly Used Modules

Python's **Standard Library** is a vast collection of modules that are installed with Python. These modules provide a rich set of functionalities for a wide range of tasks, from basic operations to complex network protocols, data serialization, and operating system interactions.

Using the standard library means you don't have to write code from scratch for common problems, saving time and ensuring robust solutions. It's often referred to as Python's "batteries included" philosophy.

Here are some of the most commonly used modules from the Python Standard Library, along with examples:

### 1. `math` Module

Provides access to mathematical functions defined by the C standard. It deals with floating-point numbers.

````python
import math

# Square root
print(f"Square root of 16: {math.sqrt(16)}")

# Pi constant
print(f"Value of Pi: {math.pi}")

# Ceiling and Floor
print(f"Ceiling of 4.3: {math.ceil(4.3)}")
print(f"Floor of 4.9: {math.floor(4.9)}")

# Factorial
print(f"Factorial of 5: {math.factorial(5)}")
````

### 2. `random` Module

Implements pseudo-random number generators for various distributions.

````python
import random

# Random integer within a range
print(f"Random integer between 1 and 10: {random.randint(1, 10)}")

# Random float between 0.0 and 1.0
print(f"Random float between 0 and 1: {random.random():.2f}")

# Random element from a sequence
my_list = ['apple', 'banana', 'cherry', 'date']
print(f"Random choice from list: {random.choice(my_list)}")

# Shuffle a list in place
random.shuffle(my_list)
print(f"Shuffled list: {my_list}")
````

### 3. `datetime` Module

Supplies classes for manipulating dates and times.

````python
from datetime import datetime, date, timedelta

# Current date and time
now = datetime.now()
print(f"Current date and time: {now}")

# Current date
today = date.today()
print(f"Current date: {today}")

# Format date and time
print(f"Formatted date and time: {now.strftime('%Y-%m-%d %H:%M:%S')}")

# Add a day to current date
tomorrow = today + timedelta(days=1)
print(f"Tomorrow's date: {tomorrow}")

# Parse a date string
date_string = "2023-10-26"
parsed_date = datetime.strptime(date_string, '%Y-%m-%d').date()
print(f"Parsed date: {parsed_date}")
````

### 4. `os` Module

Provides a way of using operating system dependent functionality. It lets you interact with the operating system, like creating directories, listing files, etc.

````python
import os

# Get current working directory
print(f"Current working directory: {os.getcwd()}")

# List contents of a directory
# print(f"Contents of current directory: {os.listdir('.')}") # Uncomment to see files

# Create a directory (if it doesn't exist)
dir_name = "my_new_directory"
if not os.path.exists(dir_name):
    os.makedirs(dir_name)
    print(f"Directory '{dir_name}' created.")
else:
    print(f"Directory '{dir_name}' already exists.")

# Check if a path exists
print(f"'my_new_directory' exists: {os.path.exists('my_new_directory')}")

# Join paths intelligently
file_path = os.path.join('data', 'reports', 'report.txt')
print(f"Joined path: {file_path}")

# Remove the created directory
# os.rmdir(dir_name) # Use os.rmdir() for empty directories
# print(f"Directory '{dir_name}' removed.")
````

### 6. `json` Module

Provides facilities for working with JSON (JavaScript Object Notation) data, which is a popular data interchange format.

````python
import json

# Python dictionary
data = {
    "name": "John Doe",
    "age": 30,
    "isStudent": False,
    "courses": [{"title": "Math", "credits": 3}, {"title": "Science", "credits": 4}]
}

# Convert Python dict to JSON string
json_string = json.dumps(data, indent=4)
print("Python dictionary converted to JSON string:")
print(json_string)

# Convert JSON string back to Python dict
parsed_data = json.loads(json_string)
print(f"\nJSON string converted back to Python dict (age): {parsed_data['age']}")
````

### 7. `collections` Module

Implements specialized container datatypes providing alternatives to Python's general purpose built-in containers (dict, list, set, tuple).

````python
from collections import Counter, defaultdict, namedtuple

# Counter: Count hashable objects
words = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']
word_counts = Counter(words)
print(f"Word counts: {word_counts}")
print(f"Most common word: {word_counts.most_common(1)}")

# defaultdict: Provides default values for missing keys
d = defaultdict(int) # Default value for int is 0
d['a'] += 1
d['b'] += 1
print(f"Defaultdict: {d}")
print(f"Value for non-existent key 'c': {d['c']}") # Accessing creates it with default value

# namedtuple: Creates tuple subclasses with named fields
Point = namedtuple('Point', ['x', 'y'])
p = Point(10, 20)
print(f"Namedtuple point: {p}")
print(f"Accessing namedtuple fields: x={p.x}, y={p.y}")
````

### 8. `re` Module

Provides regular expression operations. Regular expressions are powerful for pattern matching in strings.

````python
import re

text = "The quick brown fox jumps over the lazy dog. The fox is agile."

# Search for a pattern
match = re.search(r'fox', text)
if match:
    print(f"'fox' found at index: {match.start()}")

# Find all occurrences of a pattern
all_foxes = re.findall(r'fox', text)
print(f"All occurrences of 'fox': {all_foxes}")

# Replace a pattern
new_text = re.sub(r'fox', 'cat', text)
print(f"Text after replacement: {new_text}")

# Split a string by a pattern
words_split = re.split(r'\s+', text) # Split by one or more whitespace characters
print(f"Split into words: {words_split}")
````
