Data Manipulation and Analysis with Pandas
Data manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform, and extract insights from data. In this lesson, we will cover various data manipulation and analysis techniques using Pandas.

This is a markdown cell providing context on data manipulation and analysis with Pandas.

```python
import pandas as pd
df=pd.read_csv("/content/sales_data.csv")
```

This cell imports the pandas library, commonly aliased as `pd`, and then reads a CSV file named `sales_data.csv` from the `/content/` directory into a pandas DataFrame called `df`.

```python
df.head(5)
```

This cell displays the first 5 rows of the DataFrame `df`. This is a common practice to quickly inspect the data after loading it.

```python
df.isnull().any()
```

This code checks for the presence of any missing (NaN) values in each column of the DataFrame `df`. `isnull()` returns a boolean DataFrame indicating where values are missing, and `any()` checks if any `True` (missing) values exist in each column.

```python
df.isnull().sum()
```

This cell calculates the total count of missing (NaN) values for each column in the DataFrame `df`. As you can see from the output, there are no missing values in any column.

```python
df_filled=df.fillna(0)
```

This cell attempts to fill any missing values in the DataFrame `df` with `0`. Since the previous checks confirmed there were no missing values, this operation effectively creates a copy `df_filled` which is identical to `df`.

```python
df
```

This cell simply displays the entire DataFrame `df`.

```python
df['Sales_Amount_update']=df['Sales_Amount'].fillna(df['Sales_Amount'].mean())
df
```

This code creates a new column `Sales_Amount_update`. It attempts to fill any missing values in the `Sales_Amount` column with the mean of that column. Again, since there are no missing values in `Sales_Amount`, this operation effectively copies the `Sales_Amount` column to `Sales_Amount_update`.

```python
#rename the columns
df=df.rename(columns={"Sales_Amount" :'Sale'})
df.head()
```

This cell renames the column `Sales_Amount` to `Sale` within the DataFrame `df`. It then displays the first 5 rows of the modified DataFrame.

```python
df["Quantity_Sold_"]=df['Quantity_Sold'].fillna(df['Quantity_Sold'].mean().astype(bool))
df.head()
```

This cell attempts to create a new column `Quantity_Sold_` by filling missing values in `Quantity_Sold` with its mean, then converting that mean to a boolean type.

**Note:** While `fillna` is used, the previous checks confirmed no missing values in `Quantity_Sold`, so this operation has no practical effect on the data here. Also, converting a numerical mean to `astype(bool)` is unusual, as it would result in `True` if the mean is non-zero, and `False` if the mean is zero, which is likely not the intended way to fill numerical data. If there were missing values, `df['Quantity_Sold'].mean()` would be used directly without `astype(bool)` for a numerical fill.

```python
df['New Value']=df['Sale'].apply(lambda x:x*2)
df.head()
```

This code creates a new column named `New Value` by applying a lambda function to each value in the `Sale` column. The lambda function `lambda x: x * 2` simply doubles each sale amount. Finally, it displays the first 5 rows of the DataFrame with the new column.

```python
grouped_mean=df.groupby('Product_ID')['Sale'].mean()
print(grouped_mean)
```

This cell groups the DataFrame `df` by `Product_ID` and then calculates the mean of the `Sale` column for each `Product_ID`. The result is stored in `grouped_mean`.

```python
#
groudped_agg=df.groupby('Region')['New Value'].agg(['mean','sum','count'])
groudped_agg
```

This code groups the DataFrame `df` by `Region` and then calculates multiple aggregate statistics (mean, sum, and count) for the `New Value` column for each region. The results are stored in `groudped_agg` (note the typo in the variable name, it should ideally be `grouped_agg`).

```python
#create a new df and merging two
df1=pd.DataFrame({'key': ['A','B','C'], 'Value1':[1,2,3]})
df2=pd.DataFrame({'key': ['A','B','D'], 'Value2':[4,5,6]})
```

These lines create two new small DataFrames, `df1` and `df2`, which will be used to demonstrate merging operations. Each DataFrame has a 'key' column and a 'Value' column.

```python
df1
```

This cell displays the content of `df1`.

```python
df2
```

This cell displays the content of `df2`.

```python
pd.merge(df1,df2,on='key',how='inner')
```

This cell performs an **inner merge** of `df1` and `df2` based on the common 'key' column. An inner merge keeps only the rows where the 'key' exists in *both* DataFrames ('A' and 'B').

```python
pd.merge(df1,df2,on='key',how='outer')
```

This cell performs an **outer merge** of `df1` and `df2` based on the 'key' column. An outer merge keeps all rows from both DataFrames, filling `NaN` for values that don't have a match in the other DataFrame (e.g., 'C' from `df1` and 'D' from `df2`).

###Reading Output from Different Resource

```python
from io import StringIO
```

This line imports `StringIO` from the `io` module. `StringIO` is used to treat a string as a file-like object, which is useful when pandas functions expect a file path or buffer but you have data as a string.

```python
Data ='{"employee_name" : "James","Email" : "james@gmail.com","job_profile" : [{"title1":"team lead", "title": "Sr.Developer"}]}'
df=pd.read_json(StringIO(Data))
df
```

This cell defines a JSON string named `Data` and then uses `pd.read_json()` to parse this string into a DataFrame. `StringIO(Data)` allows `read_json` to treat the string `Data` as if it were a JSON file. Finally, it displays the resulting DataFrame.

```python
df.to_json()
```

This cell converts the DataFrame `df` (created from the JSON string) back into a JSON string using `df.to_json()`. By default, `to_json()` stores the data in a column-oriented format.

```python
df.to_json(orient='records')
```

This cell also converts the DataFrame `df` to a JSON string, but uses the `orient='records'` parameter. This format typically represents each row of the DataFrame as a separate JSON object within a list, which is often more intuitive for data exchange.
