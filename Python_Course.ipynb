## What is a Function in Python?

A function is a block of organized, reusable code that is used to perform a single, related action. Functions provide better modularity for your application and a high degree of code reusing.

**Key benefits of using functions:**
*   **Reusability:** Write code once and use it multiple times.
*   **Modularity:** Break down complex problems into smaller, manageable pieces.
*   **Readability:** Make your code easier to understand and maintain.
*   **Reduced Redundancy:** Avoid repeating the same code block.


### Defining a Function

In Python, you define a function using the `def` keyword, followed by the function name, parentheses `()`, and a colon `:`.

```python
def function_name(parameters):
    """Docstring (optional): Explains what the function does."""
    # function body - code to be executed
    return expression # (optional) returns a value
```

*   **`def`**: Keyword to declare a function.
*   **`function_name`**: A unique name for the function (follow Python's naming conventions).
*   **`parameters`**: (Optional) Values passed into the function.
*   **`:`**: Marks the end of the function header.
*   **`"""Docstring"""`**: (Optional) A string literal used to document a function's purpose. It's good practice to include one.
*   **Function Body**: Indented block of statements that the function executes.
*   **`return`**: (Optional) Keyword to send a result back to the caller. If omitted, the function implicitly returns `None`.


### Example: A Simple Greeting Function

Let's create a function that takes a name as input and returns a greeting message.

````python
# Define a function called 'greet'
def greet(name):
    """This function greets the person passed in as a parameter."""
    message = f"Hello, {name}! How are you today?"
    return message

# Call the function and store the result
my_greeting = greet('Alice')

# Display the result
display(my_greeting)

# Call the function again with a different name
display(greet('Bob'))
````

#### Explanation of the Example:

1.  **`def greet(name):`**: We define a function named `greet` that accepts one parameter, `name`.
2.  **`"""Docstring..."""`**: This is a docstring explaining what the function does.
3.  **`message = f"Hello, {name}! How are you today?"`**: Inside the function, we create a personalized greeting message using an f-string.
4.  **`return message`**: The function returns the `message` string.
5.  **`my_greeting = greet('Alice')`**: We *call* or *invoke* the `greet` function, passing the string `'Alice'` as an argument for the `name` parameter. The returned value is then assigned to the variable `my_greeting`.
6.  **`display(my_greeting)`**: We display the content of `my_greeting`.
7.  **`display(greet('Bob'))`**: We call the function again, demonstrating its reusability, and directly display the returned value.

````python
# Python Functions: Different Types of Arguments

# --- 1. Positional Arguments ---
# Arguments passed to a function in the correct positional order.
# The order of arguments matters.

def describe_person(name, age, city):
    print(f"Name: {name}, Age: {age}, City: {city}")

print("\n--- Positional Arguments Example ---")
describe_person("Alice", 30, "New York") # Arguments are matched by position
describe_person("Bob", 25, "London")

# --- 2. Keyword Arguments ---
# Arguments passed to a function using their parameter names.
# The order of keyword arguments does not matter.

def print_details(item, quantity, price):
    print(f"Item: {item}, Quantity: {quantity}, Price: ${price:.2f}")

print("\n--- Keyword Arguments Example ---")
# Arguments passed by keyword, order doesn't matter
print_details(item="Laptop", price=1200.50, quantity=1)
print_details(quantity=5, item="Pen", price=1.99)

# --- 3. Default Arguments ---
# Parameters that have a default value if no argument is provided during the function call.
# Default arguments must be placed after any non-default (positional) arguments.

def greet_user(name, message="Hello"):
    print(f"{message}, {name}!")

print("\n--- Default Arguments Example ---")
# Calling the function without providing the 'message' argument (uses default)
greet_user("Charlie")

# Calling the function and overriding the default 'message'
greet_user("David", "Good morning")

def calculate_discount(price, discount_percentage=10):
    discount_amount = price * (discount_percentage / 100)
    final_price = price - discount_amount
    print(f"Original Price: ${price:.2f}, Discount: {discount_percentage}%, Final Price: ${final_price:.2f}")

# Using the default discount percentage (10%)
calculate_discount(100)

# Providing a custom discount percentage (20%)
calculate_discount(200, 20)

print("\n--- Combined Positional, Keyword, and Default Arguments ---")
# You can combine them, but positional arguments must come first,
# followed by keyword arguments (which can include those with defaults).

def configure_settings(user, theme="dark", notifications=True):
    print(f"User: {user}, Theme: {theme}, Notifications: {notifications}")

configure_settings("User1") # Positional for 'user', defaults for 'theme' and 'notifications'
configure_settings("User2", theme="light") # Positional for 'user', keyword for 'theme', default for 'notifications'
configure_settings("User3", notifications=False, theme="blue") # Positional for 'user', keywords for 'notifications' and 'theme'
````
