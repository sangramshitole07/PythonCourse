SQL And SQLite
SQL (Structured Query Language) is a standard language for managing and manipulating relational databases. SQLite is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded database systems. In this lesson, we will cover the basics of SQL and SQLite, including creating databases, tables, and performing various SQL operations.

```python
import sqlite3
```

```python
connection=sqlite3.connect('example.bd')
connection
```

```python
cursor=connection.cursor()
```

```python
cursor.execute("CREATE TABLE IF NOT EXISTS employees(id INTEGER PRIMARY KEY, name TEXT NOT NULL, age INTEGER, department TEXT)")
```

```python
connection.commit
```

```python
cursor.execute('''
Select * from employees
    
''')
```

```python
cursor.execute(''' 
               Insert Into employees(name,age,department)
               values('krish',32,'Data Scientist')
               ''')
cursor.execute(''' 
               Insert into employees(name,age,department)
               VALUES('Bob',23,'Engineering')
               ''')

cursor.execute(''' 
               Insert into employees(name,age,department)
               values('charles',36,'Finance')
               ''')
connection.commit()
```

```python
cursor.execute("select * from employees")
row=cursor.fetchall()

for row in row:
    print(row)
```

```python
## Update the data in the table
cursor.execute('''
UPDATE employees
Set age=34
where name="Krish"
''')

connection.commit()
```

```python
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python

## Delete the data from the table
cursor.execute('''
Delete from employees
               where name ='Bob'
''')

connection.commit()
```

```python

## Query the data from the table
cursor.execute('Select * from employees')
rows=cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
    
```
