```python
import numpy as np

#creating a Array using numpy
arr=np.array([1,2,3,4,5])
print(arr)
print(type(arr))
print(arr.shape)
```

```python
#reshape array
arr2=np.array([1,2,3,4,5])
arr2.reshape(1,5)
```

```python
#or we can also declare it as 2d
arr2=np.array([[1,2,3,4,5]])
print(arr2)
print(type(arr2))
print(arr2.shape)
#properties of np/array
print(np.shape(arr2))
print(np.ndim(arr2))
print(np.size(arr2))
#functions that operate on object
type(arr2)
len(arr)
np.sum(arr)
np.mean(arr)
np.median(arr)
np.std(arr)
np.var(arr)
np.min(arr)
np.max(arr)
```
