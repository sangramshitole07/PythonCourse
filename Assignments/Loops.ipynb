```python
num=int(input("Enter a number :"))
fact=1
i=1
while i<=num:
    fact*=i
    i+=i
print(f"The factorial of {num} is {fact}.")
```

```python
#calculate the sum of the digits
num=int(input("Enter the number : "))
sum=0
while num>0:
    digit=num%10 #gives the last digit
    sum+=digit  #adds the digit
    num//=10
print(F"Sum of the digits present in the number is : {sum}")
```

```python
#prime or not 
num = int(input("Enter a number :"))
is_prime=True
if num<=0:
    is_prime=False
else :
    for i in range(2,int(num*0.5)+1):
        if num%i==0:
            is_prime=False
            break
if is_prime:
    print(f"the given  number {num} is prime")
else:
    print(f"the given number {num} is not a prime")
```
