```python
#create a list of postive number form 1 to 20
ls=list(range(1,21))
ls
```

```python
#create a list accessing first ,middle and last element
print(f" First element of the list is {ls[0]}")
print(f"Last element in the list is {ls[-1]}")
print(f"Middle element of the list is {ls[len(ls)//2]}")
```

```python
#slicing
print(f"first 5 element of the list are : {ls[:5]}")
#slicing
print(f"Last 5 element of the list are : {ls[:-5]}")
print(f" element from 5 to 15 : {ls[4:15]}")
```
