```python
students = [
    {'name': 'Alice', 'score': 88},
    {'name': 'Bob', 'score': 72},
    {'name': 'Charlie', 'score': 95},
    {'name': 'David', 'score': 65},
    {'name': 'Eve', 'score': 78}
]

sorted_student=sorted(students,key =lambda x:x['score'],reverse=True)
print(sorted_student)
print("Top students are :")
for x in sorted_student:
    print(x)
```

```python
#transpose of a matrix
def transpose(matrix):
    transposing=[[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]
    return transposing 

matrix=[[1,2,3],[4,5,6],[7,8,9]]
transposed_matrix=transpose(matrix)

print("Orginial Matrix")
for row in matrix:
    print(row)

print("For transposed Matrix")
for row in transposed_matrix:
    print(row)


```

```python
def flattedFun(nested_ls):
    flattened=[item for sublist in nested_ls for item in sublist]
    return flattened

nested_ls=[[1,2,3],[1,2,3],[12,34,54]]

result=flattedFun(nested_ls)

print("original list:")
print(nested_ls)
print("flattened list")
print(result)
```

```python
def reversed_listed(ls):
    return ls[::-1]

ls=[23,34,2,31,2134,43]
print("Original list is : ",ls)

rev_ls=reversed_listed(ls)
print('reversed list is :',rev_ls)
```
